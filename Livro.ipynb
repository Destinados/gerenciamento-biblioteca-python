```python
import matplotlib.pyplot as plt

class Livro:
    def __init__(self, titulo, autor, genero, qtd_disponivel):
        self.titulo = titulo
        self.autor = autor
        self.genero = genero
        self.qtd_disponivel = qtd_disponivel

    def __str__(self):
        return f"{self.titulo} | {self.autor} | {self.genero} | {self.qtd_disponivel} unidades"


# Lista de livros
livros = []

# Livros fixados
livros_fixados = [
    ("Dom Casmurro", "Machado de Assis", "Romance", 5),
    ("Memórias Póstumas de Brás Cubas", "Machado de Assis", "Romance", 4),
    ("Macunaíma", "Mário de Andrade", "Modernismo", 3),
    ("Triste Fim de Policarpo Quaresma", "Lima Barreto", "Pré-Modernismo", 1)
]

# Adicionando os livros fixados à lista
for titulo, autor, genero, qtd in livros_fixados:
    livros.append(Livro(titulo, autor, genero, qtd))

# cadastra novos livro e coloca na lista
def cadastrar_livros():
    qtd = int(input("Quantos livros deseja cadastrar? "))

    for _ in range(qtd):
        print("\nCadastro de novo livro:")
        titulo = input("Digite o título do livro: ")
        autor = input("Digite o autor do livro: ")
        genero = input("Digite o gênero do livro: ")
        qtd_disponivel = int(input("Digite a quantidade disponível do livro: "))

        novo_livro = Livro(titulo, autor, genero, qtd_disponivel)
        livros.append(novo_livro)

#listando todos os livros
def listar_livros():
    print("\nLISTA DE LIVROS:")
    for i, livro in enumerate(livros):
        print(f"{i+1}. {livro}")

#busca por um título específico
def buscar_livro():
    busca = input("\nDigite o título do livro: ")
    for livro in livros:
        if livro.titulo.lower() == busca.lower():
            print("\nLivro encontrado:")
            print(livro)
            return
    print("\nLivro não encontrado.")

#gráfico de barras que mostra todos os gêneros
def gerar_grafico():
    generos = {}
    for livro in livros:
        generos[livro.genero] = generos.get(livro.genero, 0) + livro.qtd_disponivel

    plt.bar(generos.keys(), generos.values())
    plt.xlabel("Gênero")
    plt.ylabel("Quantidade")
    plt.title("Quantidade de Livros por Gênero")
    plt.show()


# Exemplo de uso:
cadastrar_livros()
listar_livros()
buscar_livro()
gerar_grafico()
```

```python
import matplotlib.pyplot as plt

class Usuario:
    def __init__(self, nome, num_identificacao, contato):
        self.nome = nome
        self.num_identificacao = num_identificacao
        self.contato = contato
        self.emprestimos = []

    def __str__(self):
      return f"{self.nome} | {self.num_identificacao} | {self.contato}"

usuarios = []
emprestimos = []

def cadastrar_usuario():
    nome = input("\nNome Completo: ")
    num_identificacao = input("Número de identificação: ")

    if not num_identificacao.isdigit():
        print("Erro: ID deve ser um número inteiro.")
        return

    for usuario in usuarios:
        if usuario.num_identificacao == num_identificacao:
            print("Erro: ID já cadastrado.")
            return

    contato = input("Contato: ")

    novo_usuario = Usuario(nome, num_identificacao, contato)
    usuarios.append(novo_usuario)
    print("Usuário cadastrado com sucesso!")

def listar_usuarios():
    print("\nLISTA DE USUÁRIOS:")
    for i, usuario in enumerate(usuarios):
        print(f"{i+1}. {usuario}")

def emprestar_livro():
    print("\n=Empréstimo de Livro=")
    numero_usuario = input("ID do usuário: ")
    if not numero_usuario.isdigit():
        print("Erro: ID deve ser um número inteiro.")
        return
    livro_titulo = input("Título do livro: ")

    for usuario in usuarios:
        if usuario.num_identificacao == numero_usuario:
            for livro in livros:
                if livro.titulo == livro_titulo:
                    if livro.qtd_disponivel > 0:
                        livro.qtd_disponivel -= 1
                        emprestimo = {
                            "usuario": usuario,
                            "livro": livro
                        }
                        usuario.emprestimos.append(emprestimo)
                        emprestimos.append(emprestimo)
                        print("\n✅Empréstimo realizado com sucesso!")
                        print("Usuário: {}".format(usuario.nome))
                        print("Livro: {}".format(livro.titulo))
                        print("Quantidade disponível: {}".format(livro.qtd_disponivel))
                        return
                    else:
                        print("❌Livro indisponível para empréstimo.")
                        return
            print("❌Livro não encontrado.")
            return
    print("❌Usuário não encontrado.")

def devolver_livro():
    print("\n==Devolução de Livro==")
    numero_usuario = input("ID: ")
    livro_titulo = input("Título: ")

    for usuario in usuarios:
        if usuario.num_identificacao == numero_usuario:
            for emprestimo in usuario.emprestimos:
                if emprestimo["livro"].titulo == livro_titulo:
                    emprestimo["livro"].qtd_disponivel += 1
                    usuario.emprestimos.remove(emprestimo)
                    print("\n✅Livro devolvido com sucesso!")
                    print("Livro: {}".format(emprestimo["livro"].titulo))
                    print("Quantidade disponível: {}".format(emprestimo["livro"].qtd_disponivel))
                    return
            print("❌Livro não encontrado para devolução.")
            return
    print("❌Usuário não encontrado.")

class Livro:
    def __init__(self, titulo, autor, ano_publicacao, num_copias):
        self.titulo = titulo
        self.autor = autor
        self.ano_publicacao = ano_publicacao
        self.num_copias = num_copias
        self.qtd_disponivel = num_copias

    def __str__(self):
        return (
            f"{self.titulo} | {self.autor} | {self.ano_publicacao} |"
            f"{self.qtd_disponivel}/{self.num_copias} disponível"
        )

livros = []

livros_fixados = [
    ("Dom Casmurro", "Machado de Assis", 1899, 5),
    ("Memórias Póstumas de Brás Cubas", "Machado de Assis", 1881, 4),
    ("Macunaíma", "Mário de Andrade", 1928, 3),
    ("Triste Fim de Policarpo Quaresma", "Lima Barreto", 1927, 1)
]

for titulo, autor, ano_publicacao, num_copias in livros_fixados:
  novo_livro = Livro(titulo, autor, ano_publicacao, num_copias)
  livros.append(novo_livro)


def menu():
    print("\n==MENU==")
    print("[1]. Cadastrar Livro")
    print("[2]. Listar Livros")
    print("[3]. Buscar Livro")
    print("[4]. Gerar Gráfico")
    print("[5]. Cadastrar Usuário")
    print("[6]. Lista de Usuários")
    print("[7]. Empréstimo")
    print("[8]. Devolução")
    print("[9]. Sair")

def cadastrar_livro():
  print("\nCadastrar Livro Novo:")
  titulo = input("Título: ")
  autor = input("Autor: ")

  try:
      ano_publicacao = int(input("Ano de Publicação: "))

  except ValueError:
    print("Erro: Ano de publicação deve ser um número inteiro.")
    return

  try:
    num_copias = int(input("Número de Cópias: "))

    if num_copias <= 0:
      print("Erro: Número de cópias deve ser maior que zero.")
      return

  except ValueError:
    print("Erro: Número de cópias deve ser um número inteiro.")
    return


  novo_livro = Livro(titulo, autor, ano_publicacao, num_copias)
  livros.append(novo_livro)
  print("✅Livro cadastrado com sucesso!")

def listar_livros():
    print("\nLISTA DE LIVROS:")
    for i, livro in enumerate(livros):
        print(f"{i+1}. {livro}")

def consulta_livro(titulo):
    for livro in livros:
        if livro.titulo.lower() == titulo.lower():
          print("\nLivro encontrado:")
          print(livro)
          return livro

    return None

def gerar_grafico():

  if len(livros) == 0:
    print("Nenhum livro cadastrado.")
    return

  titulos = []
  total_copias = []
  copias_disponiveis = []
  copias_emprestadas = []

  for livro in livros:
        titulos.append(livro.titulo)
        total_copias.append(livro.num_copias)
        copias_disponiveis.append(livro.qtd_disponivel)

        emprestados = livro.num_copias - livro.qtd_disponivel
        copias_emprestadas.append(emprestados)

  x = range(len(titulos))

  plt.bar(x, total_copias, label="Total")
  plt.bar(x, copias_disponiveis, label="Disponíveis")
  plt.bar(x, copias_emprestadas, label="Emprestados")
  plt.xlabel("Livros")
  plt.ylabel("Quantidade")
  plt.title("Quantidade de Livros")
  plt.xticks(x, titulos)
  plt.tick_params(axis='x', rotation=90)
  plt.legend()
  plt.show()

while True:
      menu()

      opcao = input("Escolha uma opção: ")
      if opcao == "1":
            cadastrar_livro()
      elif opcao == "2":
            listar_livros()
      elif opcao == "3":
            titulo = input("\nDigite o título do livro: ")
            consulta_livro(titulo)
      elif opcao == "4":
            gerar_grafico()
      elif opcao == "5":
            cadastrar_usuario()
      elif opcao == "6":
            listar_usuarios()
      elif opcao == "7":
            emprestar_livro()
      elif opcao == "8":
            devolver_livro()
      elif opcao == "9":
            print("\n-Saindo do programa.-")
            break
      else:
            print("Opção inválida. Tente novamente.")
```
